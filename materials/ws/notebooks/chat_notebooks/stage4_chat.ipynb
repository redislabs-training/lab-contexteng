# Stage 4: Hybrid Search Chat

A chat interface for the Stage 4 Hybrid Search agent with ReAct reasoning.

**Features:**
- Hybrid search (semantic + exact match)
- Named Entity Recognition (NER) for course codes
- ReAct (Reasoning + Acting) loop for explicit reasoning

## Setup

Run this cell once to initialize the agent.

```python
import sys
import os
from pathlib import Path

# Map OPENAI_API_BASE to OPENAI_BASE_URL for LiteLLM compatibility
if "OPENAI_API_BASE" in os.environ:
    os.environ["OPENAI_BASE_URL"] = os.environ["OPENAI_API_BASE"]

# Enable nested event loops for Jupyter
import nest_asyncio
nest_asyncio.apply()

# Setup paths (from chat_notebooks, go up two levels to ws/)
project_root = Path("../..").resolve()
stage4_path = project_root / "progressive_agents" / "stage4_hybrid_search"
src_path = project_root / "src"
sys.path.insert(0, str(stage4_path))
sys.path.insert(0, str(src_path))

# Initialize agent
from agent import setup_agent, create_workflow
from agent.chat_interface import start_chat

print("Initializing Stage 4 Hybrid Search Agent...")
course_manager = await setup_agent(auto_load_courses=True)
workflow = create_workflow(course_manager, verbose=False)

print("Agent ready!")
```

## Start Chat

Run this cell to start chatting with the agent.

```python
start_chat(workflow)
```
