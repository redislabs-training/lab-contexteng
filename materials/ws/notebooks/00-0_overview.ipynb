# Workshop Overview: Context Engineering with Redis

Welcome to the **Context Engineering Workshop**! In this hands-on course, you'll build a Redis University Course Advisor — an AI agent that helps students discover courses, remembers their preferences, and provides personalized recommendations.

---

## 🎯 What You'll Build

Throughout this workshop, you'll progressively build a course advisor agent that demonstrates key context engineering principles:

- **Retrieval-Augmented Generation (RAG)** — Search and retrieve relevant courses from a Redis-powered knowledge base
- **Hierarchical Data Models** — Optimize token usage with progressive disclosure patterns
- **Hybrid Search** — Combine semantic and keyword search for precise results
- **Memory Systems** — Enable multi-turn conversations and cross-session personalization

By the end, you'll understand how to engineer context effectively for production AI systems.

## 🧠 The Four Context Types

This workshop is built around the **Four Context Types** framework:

| Context Type | Description | Workshop Coverage |
|--------------|-------------|-------------------|
| **System Context** | Agent identity, role, constraints | Module 0.2 |
| **User Context** | Preferences, history, personalization | Module 3.1 |
| **Conversation Context** | Multi-turn dialogue, working memory | Module 3.1 |
| **Retrieved Context** | RAG, search results, knowledge base | Modules 1.1–1.3, 2.1 |
